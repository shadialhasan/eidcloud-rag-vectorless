```python
# Cell 3: Query with Precise Citations & JSON Context Injection for LLMs
# 1. Interactive terminal output with exact line citations
!php bin/eidcloud-rag ask "How does authentication work?" --db=rag.sqlite --top=2 --citations

# 2. Clean JSON structure ready for LLM prompt context injection
!php bin/eidcloud-rag ask "Database persistence and backups" --db=rag.sqlite --top=1 --json
```
